```python
def square(n):
    return n * n

def sum_squares(x):
    total = 0
    for n in range(x):
        total += square(n)
    return total

print(sum_squares(10))
```

```python
hrs = input("Enter Hours:")
h = float(hrs)
rate = input("Enter the rate per hour:")
r = float(rate)

if h <= 40:
    pay = h * r
else:
    overtime = h - 40
    pay = (40 * r) + (overtime * r * 1.5)

print(pay)
```
