```python
# Exercise - 6
marks = [20, 30, 35, 55, 70, 80, 90]
total = 0
for m in marks:
    total += m
mean = total / len(marks)
n = len(marks)
if n % 2 == 1:
    median = marks[n // 2]
else:
    median = (marks[n // 2 - 1] + marks[n // 2]) / 2

print("Mean --> ", mean)
print("Median --> ", median)
```

```python
# Exercise - 7
def convert_to_binaryNum(n):
  binary = ""
  while n / 2 == 0:
    binary += str(n % 2)
    n = n // 2
  return binary[::-1]
print(convert_to_binaryNum(11))
```

```python
# Exercise - 8
def prodDigits(n):
  prod = 1
  while n > 0:
    prod *= n % 10
    n = n // 10
  return prod
print(prodDigits(27))
```
